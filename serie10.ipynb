### Exercise 11: Power iteration vs inverse iteration vs Rayleigh quotient iteration on matrices of fixed order with real eigenvalues

<div class="alert alert-success">
    
**Exercise 11:** Explain and run the following code in Python and comment the result.
</div>

```python
n = 1000
n_runs = 25
TimePI = numpy.zeros(n_runs)
TimeII = numpy.zeros(n_runs)
TimeRQI = numpy.zeros(n_runs)
EigErrorPI = numpy.zeros(n_runs)
EigErrorII = numpy.zeros(n_runs)
EigErrorRQI = numpy.zeros(n_runs)
for i in range(n_runs):
    P = numpy.random.randn(n, n)
    Pinv = numpy.linalg.inv(P)
    l = numpy.random.randn(n)
    A = (P*l)@Pinv
    x = numpy.random.randn(n)
    x = x/numpy.linalg.norm(x)
    TimePI[i] = time.time()
    lPI, xPI = PI(A, x)
    TimePI[i] = time.time()-TimePI[i]
    EigErrorPI[i] = min(abs(l-lPI))
    TimeII[i] = time.time()
    lII, xII = II(A, x, x.dot(A.dot(x)))
    TimeII[i] = time.time()-TimeII[i]
    EigErrorII[i] = min(abs(l-lII))
    TimeRQI[i] = time.time()
    lRQI, xRQI = RQI(A, x)
    TimeRQI[i] = time.time()-TimeRQI[i]
    EigErrorRQI[i] = min(abs(l-lRQI))
print('Run time of PI: ', [min(TimePI), numpy.median(TimePI), max(TimePI)])
print('Run time of II: ', [min(TimeII), numpy.median(TimeII), max(TimeII)])
print('Run time of RQI: ', [min(TimeRQI), numpy.median(TimeRQI), max(TimeRQI)])
print('Error on the eigenvalue committed by PI: ', [min(EigErrorPI), numpy.median(EigErrorPI), max(EigErrorPI)])
print('Error on the eigenvalue committed by II: ', [min(EigErrorII), numpy.median(EigErrorII), max(EigErrorII)])
print('Error on the eigenvalue committed by RQI: ', [min(EigErrorRQI), numpy.median(EigErrorRQI), max(EigErrorRQI)])
```

### Exercise 15: QR algorithm

<div class="alert alert-success">
    
**Exercise 15:** Explain and run the following code in Python and comment the result.
</div>

```python
# QR factorization
def QR(A):
    Q, R = numpy.linalg.qr(A)
    s = numpy.sign(numpy.diag(R))
    Q = Q*s
    R = numpy.transpose(numpy.transpose(R)*s)
    return Q, R

# QR iteration
def QRI(A):
    for i in range(100):
        Q, R = numpy.linalg.qr(A)
        A = R@Q
    return A

A = numpy.array([[3, 2, 2, -4], [2, 3, 2, -1], [1, 1, 2, -1], [2, 2, 2, -1]])
print(QRI(A))
```
